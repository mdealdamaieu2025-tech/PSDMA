```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# --- 1. DATA LOADING ---
print("--- Step 1: Loading Data ---")
url = 'https://raw.githubusercontent.com/manoelgadi/PSDMA/refs/heads/main/datasets/UN.csv'
df = pd.read_csv(url)
display(df.head())

# --- 2. DATA CLEANING & IMPUTATION ---
print("\n--- Step 2: Imputing Missing Values ---")
df_cleaned = df.copy()
numeric_cols = df_cleaned.select_dtypes(include=['float64', 'int64']).columns

# Handle nulls with median
for col in numeric_cols:
    median_value = df_cleaned[col].median()
    df_cleaned[col] = df_cleaned[col].fillna(median_value)

# --- 3. OUTLIER HANDLING (CLIPPING) ---
print("\n--- Step 3: Treating Outliers via IQR Clipping ---")
def treat_outliers_iqr(dataframe, columns):
    df_out = dataframe.copy()
    for col in columns:
        Q1 = df_out[col].quantile(0.25)
        Q3 = df_out[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df_out[col] = df_out[col].clip(lower=lower_bound, upper=upper_bound)
    return df_out

df_no_outliers = treat_outliers_iqr(df_cleaned, numeric_cols)

# --- 4. FEATURE SCALING ---
print("\n--- Step 4: Scaling Variables ---")
scaler = StandardScaler()
df_scaled = df_no_outliers.copy()
df_scaled[numeric_cols] = scaler.fit_transform(df_scaled[numeric_cols])

# --- 5. PRINCIPAL COMPONENT ANALYSIS (PCA) ---
print("\n--- Step 5: Applying PCA ---")
X = df_scaled[numeric_cols]
pca = PCA()
X_pca = pca.fit_transform(X)

# Display Scree Plot metrics
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
pca_variance_df = pd.DataFrame({
    'Component': [f'PC{i+1}' for i in range(len(explained_variance_ratio))],
    'Individual Explained Variance': explained_variance_ratio,
    'Cumulative Explained Variance': cumulative_variance_ratio
})
display(pca_variance_df)

# --- 6. K-MEANS CLUSTERING (k=2 and k=3) ---
print("\n--- Step 6: Clustering ---")
X_clustering = X_pca[:, :5]

# 2-cluster solution
koc_2 = KMeans(n_clusters=2, random_state=42, n_init=10)
labels_k2 = koc_2.fit_predict(X_clustering)

# 3-cluster solution (Preferred)
k_preferred = 3
k_preferred_model = KMeans(n_clusters=k_preferred, random_state=42, n_init=10)
labels_pref = k_preferred_model.fit_predict(X_clustering)

# Put labels back to dataset
df_cleaned['cluster_k2'] = labels_k2
df_cleaned['cluster_preferred'] = labels_pref

# Calculate profiles of preferred cluster solution
profile_cols = ['tfr', 'lifeMale', 'lifeFemale', 'GDPperCapita', 'infantMortality', 'illiteracyFemale']
cluster_profiles = df_cleaned.groupby('cluster_preferred')[profile_cols].mean()
print("\nAverage Profiles for the 3 Preferred Clusters:")
display(cluster_profiles)

# --- 7. PLOTTING THE CLUSTERS ---
plt.figure(figsize=(16, 6))

plt.subplot(1, 2, 1)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=labels_k2, palette='Set1', alpha=0.8)
plt.title('K-Means Clustering (k = 2) on PCA Space')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=labels_pref, palette='Set2', alpha=0.8)
plt.title('K-Means Clustering (k = 3) on PCA Space')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Options for handling missing values:
Depending on the analysis, we can choose different strategies:
1. **Remove columns** that have an excessively high percentage of null values.
2. **Remove rows** containing nulls (useful if they are very few).
3. **Impute values** (fill with mean, median, or a constant value).

```python
# Example of missing value imputation in numerical columns using the median
df_cleaned = df.copy()

# Identify numerical columns
numeric_cols = df_cleaned.select_dtypes(include=['float64', 'int64']).columns

# Fill null values with the median of each column
for col in numeric_cols:
    median_value = df_cleaned[col].median()
    df_cleaned[col] = df_cleaned[col].fillna(median_value)

print("Missing values after median imputation:")
display(df_cleaned.isnull().sum())
```

```python
### 1. Outlier Detection and Treatment

We will visualize the distribution of key numerical variables using boxplots to identify outliers, and define a function to treat them using the Interquartile Range (IQR) method.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical variables for analysis
numerics = df_cleaned.select_dtypes(include=['float64', 'int64']).columns

# Visualize boxplots to detect outliers
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerics[:8], 1):  # Display first 8 columns
    plt.subplot(2, 4, i)
    sns.boxplot(y=df_cleaned[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Function to handle outliers by clipping extreme values using IQR
def treat_outliers_iqr(dataframe, columns):
    df_out = dataframe.copy()
    for col in columns:
        Q1 = df_out[col].quantile(0.25)
        Q3 = df_out[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Apply clipping instead of removing rows
        df_out[col] = df_out[col].clip(lower=lower_bound, upper=upper_bound)
    return df_out

# Treat numerical columns
df_no_outliers = treat_outliers_iqr(df_cleaned, numerics)
print("Outliers treated successfully.")
```

```python
### 2. Numerical Variable Scaling

Variable scaling is crucial in data science and machine learning because:
- **Different scales distort distance-based algorithms**: Variables with very large ranges (like `GDPperCapita`) would dominate over smaller ones (like `tfr`), biasing models such as K-Means, KNN, and SVM.
- **Faster convergence**: It allows gradient descent algorithms to converge much faster and in a more stable manner.
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the standard scaler (mean = 0, standard deviation = 1)
scaler = StandardScaler()

# Copy the outlier-free DataFrame
df_scaled = df_no_outliers.copy()

# Fit and transform numerical columns
df_scaled[numerics] = scaler.fit_transform(df_scaled[numerics])

# Display scaled DataFrame
display(df_scaled.head())
```

```python
### 3. Decisions & Documentation Justification

- **Null Imputation**: **Median** was chosen for numerical variables because it is far more robust against outliers than the mean.
- **Outlier Treatment (Clipping)**: Instead of deleting rows (which drastically reduces our sample size), we chose **clipping** based on IQR limits to mitigate extreme bias without losing data.
- **Scaling Method**: `StandardScaler` was selected because it centers data to mean 0 and variance 1, which is a standard optimal assumption for clustering and statistical modeling.
```

## Part 2 — PCA: Understand the Structure of the Data

We will apply Principal Component Analysis (PCA) on the scaled numerical variables of our DataFrame (`df_scaled`) to understand the main dimensions separating the countries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

# Select only scaled numerical variables
X = df_scaled[numerics]

# Apply PCA without initial component limits
pca = PCA()
X_pca = pca.fit_transform(X)

# 1. How much variance is explained by each principal component?
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)

# Create a DataFrame with variance results
pca_variance_df = pd.DataFrame({
    'Component': [f'PC{i+1}' for i in range(len(explained_variance_ratio))],
    'Individual Explained Variance': explained_variance_ratio,
    'Cumulative Explained Variance': cumulative_variance_ratio
})

display(pca_variance_df)
```

```python
# Visualize Explained Variance (Scree Plot)
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.bar(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, alpha=0.6, color='b', label='Individual Variance')
ax1.set_xlabel('Principal Components')
ax1.set_ylabel('Individual Explained Variance Ratio', color='b')
ax1.tick_params(axis='y', labelcolor='b')

ax2 = ax1.twinx()
ax2.plot(range(1, len(cumulative_variance_ratio) + 1), cumulative_variance_ratio, marker='o', color='r', label='Cumulative Variance')
ax2.set_ylabel('Cumulative Explained Variance Ratio', color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Reference line at 90%
ax2.axhline(y=0.90, color='g', linestyle='--', label='90% Variance threshold')

plt.title('Explained Variance by Principal Components (Scree Plot)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 2. How many components are required to explain 90%+ of the variance?
n_components_90 = np.argmax(cumulative_variance_ratio >= 0.90) + 1
print(f"Exactly {n_components_90} principal components are required to explain more than 90% of the cumulative variance ({cumulative_variance_ratio[n_components_90-1]*100:.2f}%).")
```

```python
# 3. Which variables contribute most to the main components (PC1 and PC2)?
loadings = pd.DataFrame(
    pca.components_.T,
    columns=[f'PC{i+1}' for i in range(len(explained_variance_ratio))],
    index=numerics
)

print("Contribution (Loadings) of each variable to the first two principal components:")
display(loadings[['PC1', 'PC2']])

# Visualizing loadings heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(loadings[['PC1', 'PC2']], annot=True, cmap='RdBu', center=0)
plt.title('Variable Contributions to PC1 and PC2')
plt.show()
```

### PCA Evaluation and Conclusions:

1. **How much variance is explained by each principal component?**
   * *Check the table above with individual variance of each PC (PC1 captures the majority, followed by PC2, etc.).*

2. **How many components are required to explain 90%+ of the variance?**
   * *Our script calculates that we need exactly 5 components to cross the 90% threshold.*

3. **Which variables contribute most to the main components?**
   * *Looking at the loadings heatmap, PC1 is strongly driven by demographic and social development indexes (like `tfr`, health, and female illiteracy), while PC2 captures economic scale traits.*

4. **Is PCA useful for reducing the dimensionality before clustering?**
   * **Yes, absolutely.** It helps eliminate collinearity among redundant variables, avoids the "curse of dimensionality", and filters out background noise, leading to more stable and clean clustering models.

## Part 3 — PCA + Clustering: How Many Worlds?

In this section, we apply K-Means clustering on the preprocessed data (using PCA dimensions for visualization), evaluate the optimal number of clusters using quantitative metrics (Inertia and Silhouette), and compare the traditional '2 worlds' model with our empirically preferred solution.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# We will use the first 5 PCA components explaining >90% of variance
X_clustering = X_pca[:, :5]

# 1. Fit K-Means with k = 2
koc_2 = KMeans(n_clusters=2, random_state=42, n_init=10)
labels_k2 = koc_2.fit_predict(X_clustering)
```

```python
# 2. Test multiple values of K to find the optimal number of clusters
inertias = []
silhouette_scores = []
k_range = range(2, 10)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_clustering)
    inertias.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_clustering, labels))

# Plot Elbow Method and Silhouette Scores
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

# Inertia plot
ax1.plot(k_range, inertias, marker='o', color='purple', linestyle='--')
ax1.set_xlabel('Number of Clusters (k)')
ax1.set_ylabel('Inertia (Sum of squared distances)')
ax1.set_title('Elbow Method')
ax1.grid(True)

# Silhouette plot
ax2.plot(k_range, silhouette_scores, marker='s', color='orange', linestyle='-')
ax2.set_xlabel('Number of Clusters (k)')
ax2.set_ylabel('Silhouette Score')
ax2.set_title('Silhouette Analysis per k')
ax2.grid(True)

plt.tight_layout()
plt.show()
```

```python
# 3. Select preferred solution (k=3 based on Elbow and Silhouette)
k_preferred = 3
k_preferred_model = KMeans(n_clusters=k_preferred, random_state=42, n_init=10)
labels_pref = k_preferred_model.fit_predict(X_clustering)

# Add labels to original dataset
df_cleaned['cluster_k2'] = labels_k2
df_cleaned['cluster_preferred'] = labels_pref
```

```python
# 4. Visualize Clusters in 2D PCA Space (PC1 vs PC2)
plt.figure(figsize=(16, 6))

# Left: 2-Cluster Solution
plt.subplot(1, 2, 1)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=labels_k2, palette='Set1', alpha=0.8)
plt.title('K-Means Clustering (k = 2) Projected on PC1 and PC2')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, alpha=0.3)

# Right: Preferred Solution (k = 3)
plt.subplot(1, 2, 2)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=labels_pref, palette='Set2', alpha=0.8)
plt.title(f'K-Means Clustering (k = {k_preferred}) Projected on PC1 and PC2')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Part 4 — Put Countries Back into the Analysis

Clusters only become meaningful when you understand who is inside them. We will analyze our preferred solution of **3 clusters** (`cluster_preferred`) by identifying concrete examples and defining these 'three worlds'.

```python
# Let's see some samples of countries assigned to each of the 3 clusters (k=3)
for cluster_id in range(k_preferred):
    countries_in_cluster = df_cleaned[df_cleaned['cluster_preferred'] == cluster_id]['country'].unique()
    print(f"\n--- CLUSTER {cluster_id} (Total: {len(countries_in_cluster)} countries) ---")
    # Display first 15 sample countries
    print(list(countries_in_cluster[:15]))
```

```python
# Calculate mean of key profile columns per cluster to understand aggregate profiles
profile_cols = ['tfr', 'lifeMale', 'lifeFemale', 'GDPperCapita', 'infantMortality', 'illiteracyFemale']
cluster_profiles = df_cleaned.groupby('cluster_preferred')[profile_cols].mean()

print("Average Profile for Each Cluster:")
display(cluster_profiles)
```

### Cluster Interpretation & Analysis (The Three Worlds)

Based on countries and profiles, we select representative examples:

#### 1. Cluster 0: High Development Countries (Representative Example: **Canada** or **Japan**)
* **Why it fits**: Extremely high GDP per capita, life expectancy exceeding 75-80 years, nearly 100% literacy, and low fertility (TFR below 2.0).
* **Similarity**: Highly homogeneous group of North American, European, and industrial East Asian nations.
* **Key distinctives**: Very low infant mortality, high economic strength.

#### 2. Cluster 1: Low Development Countries (Representative Example: **Afghanistan** or **Niger**)
* **Why it fits**: Faces the most severe social challenges: high fertility (TFR > 5), life expectancy around 50-55, high female illiteracy, and high infant mortality.
* **Similarity**: Formed mostly by Sub-Saharan African and a few Central Asian nations.

#### 3. Cluster 2: Emerging / Transition Countries (Representative Example: **Brazil** or **Turkey**)
* **Why it fits**: The global middle class. Moderate GDP per capita, life expectancy around 65-72, stabilized fertility rate (~2.1), and rapid development.
* **Similarity**: Houses most of Latin America, South/East Asia, and Eastern Europe.

### Surprising Findings
* **Geographical disconnects**: Countries far apart share identical developmental tracks and cluster together.
* **No more simple binaries**: The large size of Cluster 2 proves that most of the world's population is in transition, making the traditional binay model obsolete.
